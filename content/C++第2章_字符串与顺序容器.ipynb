## 1. std::string 管理文本

字符串可以赋值、拼接、比较；使用 `==` 比较内容。`size()` 是字符单元数，UTF-8 中文通常占多个字节，不能当作中文字符个数。下面使用 ASCII 姓名便于观察。`find` 未找到时返回 `std::string::npos`，不是0。

````c++
#include <iostream>
#include <string>
{
    const std::string name{"Alice"};
    const std::string report = name + ": pass";
    std::cout << report << '\n' << name.size() << ' ' << (name.find('z') == std::string::npos) << '\n';
}
````

## 2. 固定长度 array 与可变长度 vector

`std::array<int, 3>` 的元素个数属于类型；`std::vector<int>` 可动态增长，自动管理内存。`size()` 是已有元素数，`capacity()` 是容量；`reserve` 不创建元素，`resize` 才改变元素数。`[]` 不检查越界，`at()` 越界会抛出异常。先判断 empty，再取 front/back。

**预测**：reserve 后size是多少？把 push_back 的分数改成0再运行。

````c++
#include <array>
#include <vector>
#include <iostream>
{
    const std::array<int, 3> seed{85, 90, 58};
    std::vector<int> scores(seed.begin(), seed.end());
    scores.reserve(20);
    std::cout << scores.size() << '\n';
    scores.push_back(60);
    int total{};
    for (const int score : scores) { total += score; }
    std::cout << scores.size() << ' ' << total << '\n';
}
````

## 3. 值语义与范围 for

复制 vector 会得到独立的元素副本。`auto value` 拷贝元素；`auto& value` 才能通过引用修改容器元素；只读复杂元素用 `const auto&`。此处演示修改局部副本，原输入保持不变。扩容可能使原来的指针、引用、迭代器失效，不能保存后跨 push_back 继续使用；需要时重新获取。

````c++
#include <vector>
#include <iostream>
{
    const std::vector<int> original{58, 90};
    auto adjusted = original;
    for (auto& value : adjusted) { value += 2; }
    std::cout << original.at(0) << ' ' << adjusted.at(0) << '\n';
}
````

## 本地实践：完整 C++17 程序

把代码保存为 `lesson.cpp`，在终端执行 `c++ -std=c++17 -Wall -Wextra -Wpedantic lesson.cpp -o lesson`，编译成功后运行 `./lesson`。Windows GCC 使用 `g++` 和 `-o lesson.exe`，PowerShell 运行 `.\lesson.exe`。完整程序不要直接放入在线单元或手动调用 main。

```cpp
#include <iostream>
#include <vector>
int main() {
    int count{};
    if (!(std::cin >> count) || count < 0 || count > 100) { std::cout << "invalid\n"; return 1; }
    std::vector<int> scores;
    for (int i = 0; i < count; ++i) {
        int score{};
        if (!(std::cin >> score) || score < 0 || score > 100) { std::cout << "invalid\n"; return 1; }
        scores.push_back(score);
    }
    int passed{};
    for (const int score : scores) { if (score >= 60) { ++passed; } }
    std::cout << "count=" << scores.size() << " passed=" << passed << '\n';
}
```

验收输入与输出（`\n` 表示换行）：

- 输入 `"3 85 90 58"` → 输出 `"count=3 passed=2\n"`，退出码 0。
- 输入 `"0"` → 输出 `"count=0 passed=0\n"`，退出码 0。
- 输入 `"2 60"` → 输出 `"invalid\n"`，退出码 1。
- 输入 `"1 101"` → 输出 `"invalid\n"`，退出码 1。
