## 1. 同样的成绩报告，不同的表达

C++ 是独立的语言，不是把所有 C 文件改后缀即可。C++ 支持类与泛型；本课程先使用标准库管理文本和集合。`std::` 指定标准库命名空间，避免全局 `using namespace std;`。`<<` 在输出流中表示插入。`\n` 换行，通常不需要每行用 std::endl 强制刷新。

**预测**：整数除法是否改变？修改两项分数，再解释转换为何在除法前发生。

````c++
#include <iostream>
#include <string>
{
    const std::string name{"Alice"};
    const int total{85 + 90};
    const double average = static_cast<double>(total) / 2;
    std::cout << name << ": " << average << '\n';
}
````

## 2. 引用与值拷贝

`int&` 是已有对象的别名，定义时必须绑定对象，之后不能改绑；给引用赋值会修改原对象。值拷贝则拥有独立的值。`const T&` 常用于只读访问较大的对象，小整数直接传值即可。不要返回局部变量的引用或指针。

**预测**三个值，再把 copy 的赋值改为其他数字。

````c++
#include <iostream>
{
    int score{58};
    int copy = score;
    int& alias = score;
    alias = 60;
    copy = 90;
    const int& view = score;
    std::cout << score << ' ' << copy << ' ' << view << '\n';
}
````

## 3. const、constexpr、auto 与重载

`const` 表示不能通过该名字修改对象，不保证值在编译期可知；`constexpr` 变量要求常量表达式初始化。`auto` 从初始化表达式推导类型，不表示动态类型。花括号初始化能拒绝一些窄化转换：`int n{3.5};` 应编译失败，只在副本中试错。

函数重载按参数区分，不能只靠返回类型区分；本例中 3 调用 int 版本，3.0 调用 double 版本。默认参数如 `int bonus(int score, int extra = 0)` 在省略尾部实参时使用默认值，但不要设计容易产生歧义的重载。

````c++
#include <iostream>
int twice_value(int value) { return value * 2; }
double twice_value(double value) { return value * 2; }
````

````c++
{
    constexpr int pass_line{60};
    const auto passed = twice_value(30) >= pass_line;
    std::cout << twice_value(3) << ' ' << twice_value(3.5) << ' ' << passed << '\n';
}
````

## 4. 输入失败也是程序的一条路径

在线使用字符串流练习提取；它与键盘输入来源不同。`if (input >> score)` 检查转换是否成功，还要检查业务范围。初始化避免读取未赋值变量。这里要求只有一个整数，额外字符也拒绝。

````c++
#include <sstream>
#include <iostream>
#include <string>
{
    for (const std::string text : {"85", "abc", "101", "42abc"}) {
        std::istringstream input{text};
        int score{};
        std::string extra;
        const bool valid = (input >> score) && score >= 0 && score <= 100 && !(input >> extra);
        std::cout << text << ": " << (valid ? "valid" : "invalid") << '\n';
    }
}
````

## 本地实践：完整 C++17 程序

把代码保存为 `lesson.cpp`，在终端执行 `c++ -std=c++17 -Wall -Wextra -Wpedantic lesson.cpp -o lesson`，编译成功后运行 `./lesson`。Windows GCC 使用 `g++` 和 `-o lesson.exe`，PowerShell 运行 `.\lesson.exe`。完整程序不要直接放入在线单元或手动调用 main。

```cpp
#include <iostream>
#include <sstream>
#include <string>
int main() {
    std::string line;
    if (!std::getline(std::cin, line)) { std::cout << "invalid\n"; return 1; }
    std::istringstream input{line};
    int score{};
    std::string extra;
    if (!(input >> score) || score < 0 || score > 100 || (input >> extra)) {
        std::cout << "invalid\n"; return 1;
    }
    std::cout << (score >= 60 ? "pass" : "fail") << '\n';
}
```

验收输入与输出（`\n` 表示换行）：

- 输入 `"60\n"` → 输出 `"pass\n"`，退出码 0。
- 输入 `"59\n"` → 输出 `"fail\n"`，退出码 0。
- 输入 `"42abc\n"` → 输出 `"invalid\n"`，退出码 1。
- 输入 `"999999999999999999\n"` → 输出 `"invalid\n"`，退出码 1。
- 输入 `""` → 输出 `"invalid\n"`，退出码 1。
